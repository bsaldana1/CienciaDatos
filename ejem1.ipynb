# Ejemplo de estadistica descriptiva

```python
import pandas as pd
```

```python
datos = pd.Series([70,80,90,100,69,55,77,85,94])

print(datos.mean())
print(datos.median())
print(datos.mode())
print(datos.std())


```

# Ejemplo de graficas  

```python
# grafica de lineal

import matplotlib.pyplot as plt

ventas = [100, 120, 150, 180]

plt.plot(ventas)
plt.title("Ventas mensuales")
plt.show()
```

```python
# grafica de barras
productos = ["A","B", "C", "D"]

plt.bar(productos, ventas)
plt.show()
```

```python
# grafica histograma
plt.hist(ventas)
plt.show()
```

```python
plt.hist([60,65,70,75,80,85,90,95])
plt.show()
```

```python
import seaborn as sns
import pandas as pd

datos = pd.DataFrame(
    ventas,
    productos
)
```

```python
sns.scatterplot(
    data= datos,
    y=ventas,
    x= productos
)
```

```python
# boxplot
sns.boxplot(data=datos)
```

```python
# heatmap
sns.heatmap(datos.corr(), annot=True)
```

```python
sns.heatmap(datos.corr(), annot=False)
```

```python
plt.style.use('seaborn-v0_8')
```

```python
plt.Figure(figsize=(10,6))

plt.plot(productos, ventas, marker='o', linestyle='-', color='green', label='Rendimiento')

plt.title("Tendencia de ventas por producto", fontsize=14, fontweight='bold')
plt.xlabel('Producto', fontsize=12)
plt.ylabel('Ventas',fontsize=12)

plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

plt.show()
print(datos)
```

```python
#Grafico de dispersion
import pandas as pd
import plotly.express as px
```

```python
datos = pd.DataFrame({
     "Horas": [2,4,5,6,7,8,12],
    "Calificacion":[60,70,55,80,95,100,90]
})
```

```python
#Grafica de dispersion ( Scatter)
fig1 = px.scatter(
    datos,
    x="Horas",
    y="Calificacion",
    title="Relacion entre estudio y calificacion",
    labels={"Horas":"Horas de estudio", "Calificacion":"Calificacion"},
    size="Calificacion",
    color="Horas",
    color_continuous_scale="viridis"
)

fig1.show()
```

```python
#Grafica de barras usando los mismos datos
fig2 = px.bar(
    datos,
    x="Horas",
    y="Calificacion",
    title="Relacion entre estudio y calificacion",
    labels={"Horas":"Horas de estudio", "Calificacion":"Calificacion"},
    color="Horas",
    color_continuous_scale="viridis"
)

fig2.show()
```

```python
#Grafica de histograma usando los mismos datos
fig3 = px.histogram(
    datos,
    x="Horas",
   
    title="Relacion entre estudio y calificacion",
    labels={ "Calificacion":"Calificacion"},
    nbins=10
)

fig3.show()
```

```python
#Grafica de barras usando los mismos datos
fig4 = px.box(
    datos,
    x="Horas",
    y="Calificacion",
    title="Relacion entre estudio y calificacion",
    labels={"Horas":"Horas de estudio", "Calificacion":"Calificacion"},
    color="Horas"
    
)

fig4.show()
```

```python
#Grafica de barras usando los mismos datos
fig5 = px.bar_polar(
    datos,
    
   
    title="Relacion entre estudio y calificacion",
    labels={"Horas":"Horas de estudio", "Calificacion":"Calificacion"},
    color="Horas"
)

fig5.show()
```
