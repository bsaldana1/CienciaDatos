# Practica 7 trabajando con graficas y correlacion

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Configuracion para mostrar graficos en jupyther
%matplotlib inline
```

```python
# Datos ejemplo
datos =  pd.DataFrame({
     "Horas": [2,4,5,6,7,8,12],
    "Calificacion":[60,70,55,80,95,100,90]
})
```

```python
# Mostras correlacion
print("La correlacion de datos es ", datos.corr(),"\n")
```

```python
# Gafrico de dispersion (Scatter Plot)
plt.Figure(figsize=(10,6))
plt.scatter(datos['Horas'],datos['Calificacion'],
            color='blue', s=100 ,alpha=0.7, edgecolors='black')

# Añadir lineas de tendencia
z = np.polyfit(datos['Horas'],datos['Calificacion'],1)
p = np.poly1d(z)
plt.plot(datos['Horas'],p(datos['Horas']),
         color='red', linestyle='-', linewidth=2,label=f"Tendecia : {z[0]:.2f} x + {z[1]:.2f}")

# personalizar
plt.title("Relacion entre hora de estudio y califiacion", fontsize=16, fontweight='bold')
plt.xlabel("Horas de estudio", fontsize=12)
plt.ylabel("Calificacion", fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Gafrico de lineas
plt.Figure(figsize=(10,6))
plt.plot(datos['Horas'],datos['Calificacion'],
         marker='o',linestyle='-', linewidth=1, markersize=5,
            color='green', markerfacecolor="yellow",markeredgewidth=2)

# Añadir etiquetas con valores
for i, (h , c) in enumerate(zip(datos['Horas'],datos['Calificacion'])):
    plt.annotate(f'({h}h, {c}pts)',
                (h,c),
                textcoords="offset points",
                xytext=(10,10),
                ha="center",
                fontsize=10,
                bbox=dict(boxstyle="round, pad=0.3",facecolor='gray', alpha=0.3))
    

# personalizar
plt.title("Relacion entre hora de estudio y califiacion", fontsize=16, fontweight='bold')
plt.xlabel("Horas de estudio", fontsize=12)
plt.ylabel("Calificacion", fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(datos['Horas'])
plt.yticks(range(50,101,10))
plt.show()
```

```python
# Grafico de area
plt.figure(figsize=(10,6))
plt.fill_between(datos['Horas'], datos['Calificacion'],
                 color="pink", alpha=0.3, label="Area bajo la curva")
plt.plot(datos['Horas'], datos['Calificacion'],
         marker='o', linewidth=2, color='darkblue', label="Calificacion")

plt.title("Progreso de calificacion por horas de estudio", fontsize=16, fontweight='bold')
plt.xlabel("Horas de estudio", fontsize=12)
plt.ylabel("Calificacion", fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.xticks(datos['Horas'])
plt.yticks(range(50,101,10))
plt.show()
```

```python
# Grafico combinado
fig , axes  = plt.subplots(2,2, figsize=(12,10))

# scatter
axes[0,0].scatter(datos['Horas'], datos['Calificacion'], color='purple', s=80)
axes[0,0].set_title('Dispersion', fontweight='bold')
axes[0,0].grid(True, alpha=0.3)

# lineas
axes[0,1].plot(datos['Horas'], datos['Calificacion'], marker='o' ,color='orange')
axes[0,1].set_title('Lineas', fontweight='bold')
axes[0,1].grid(True, alpha=0.3)

# bar
axes[1,0].bar(datos['Horas'].astype(str), datos['Calificacion'], color='teal')
axes[1,0].set_title('Barras', fontweight='bold')
axes[1,0].grid(True, alpha=0.3, axis='y')

# area
axes[1,1].fill_between(datos['Horas'], datos['Calificacion'], alpha=0.5, color="coral")
axes[1,1].plot(datos['Horas'], datos['Calificacion'], marker='o' ,color='darkred')
axes[1,1].set_title('area', fontweight='bold')
axes[1,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Resumen de correlacion
corr_value = datos.corr().iloc[0,1]
print("Resumen")
print(f"Correlacion entre horas y calificacion: {corr_value:.4f}")
print(f"Interpretacion: Correlacion {'Muy Fuerte' if corr_value>0.9 else 'Fuerte' if corr_value>0.7 else 'Moderada' if corr_value>0.5 else 'Devil'}")
print("A mas horas de estudio, mayor calificacion(Relacion positiva)")
```
